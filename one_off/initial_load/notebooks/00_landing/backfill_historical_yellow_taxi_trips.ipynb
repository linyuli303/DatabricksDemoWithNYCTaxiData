```python
import urllib.request
import shutil
import os
```

```python
# Initial data includes the first 6 months of 2026; July and August 2026 will be added by leveraging Slowly Changing Dimension Type 2 (SCD Type 2)
data_to_process = ["2026-01", "2026-02", "2026-03", "2026-04", "2026-05", "2026-06", "2026-07"]
for date in data_to_process:
    url = f"https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_{date}.parquet"
    response = urllib.request.urlopen(url)
    dir_path = f"/Volumes/yuli_demo_nyctaxi/00_landing/data_sources/nyctaxi_yellow/{date}"
    os.makedirs(dir_path,exist_ok="True")
    file_path = f"{dir_path}/yellow_tripdata_{date}.parquet"
    with open(file_path, 'wb') as f:
        shutil.copyfileobj(response, f)
```
