```python
from pyspark.sql.functions import *
```

```python
spark.read.table("yuli_demo_nyctaxi.01_bronze.yellow_trips_raw").\
    groupBy(date_format("tpep_pickup_datetime", "yyyy-MM").alias("year_month")).\
    agg(count("*").alias("total_records")).\
    orderBy("year_month").display()
```

```python
spark.read.table("yuli_demo_nyctaxi.02_silver.yellow_trips_cleansed").\
    groupBy(date_format("tpep_pickup_datetime", "yyyy-MM").alias("year_month")).\
    agg(count("*").alias("total_records")).\
    orderBy("year_month").display()
```

```python
spark.read.table("yuli_demo_nyctaxi.02_silver.yellow_trips_enriched").\
    groupBy(date_format("tpep_pickup_datetime", "yyyy-MM").alias("year_month")).\
    agg(count("*").alias("total_records")).\
    orderBy("year_month").display()
```
